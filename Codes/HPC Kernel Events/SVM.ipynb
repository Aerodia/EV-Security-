```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
data = pd.read_csv("EVSE-B-PowerCombined_Cleaned.csv")
data.head()
```

```python


# # Load the dataset
# data = pd.read_csv('EVSE-B-PowerCombined_Cleaned.csv')  # Replace with your file path

# Preprocess the data: drop irrelevant columns and one-hot encode categorical features
data_cleaned = data.drop(columns=['time'])  # Drop irrelevant column
data_cleaned = pd.get_dummies(data_cleaned, drop_first=True)  # One-hot encoding

# Define features (X) and target variable (y)
X = data_cleaned.drop(columns=['power_mW'])  # Features
y = (data_cleaned['power_mW'] > data_cleaned['power_mW'].mean()).astype(int)  # Binary classification (0 or 1)

# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train the SVM model (with linear kernel for simplicity)
model = SVC(kernel='linear', random_state=42)
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate the model's performance
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

# Print results
print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
```
